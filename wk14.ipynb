### • Принимать три параметра (a: int, b: int, length: int) и возвращать список (list) длины length из равномерно распределенной случайной величины между числами a и b

```python
import random
from collections import Counter
import statistics

def generate_uniform_list(a: int, b: int, length: int) -> list:
    return [random.randint(a, b) for _ in range(length)]
```

### • Принимать созданный список (list) и считать количество уникальных значений переменной (строить частотный ряд) в виде dict. Например, для списка [1, 2, 3, 1, 2, 2] будет возвращать {1 : 2, 2 : 3, 3 : 1}

```python
def frequency_dict(lst: list) -> dict:
    return dict(Counter(lst))
```

### • Считать минимум, среднюю, медиану, моду, максимум для значений списка

```python
def describe_list(lst: list) -> dict:
    try:
        mode_value = statistics.mode(lst)
    except statistics.StatisticsError:
        mode_value = None

    return {
        "min": min(lst),
        "mean": statistics.mean(lst),
        "median": statistics.median(lst),
        "mode": mode_value,
        "max": max(lst),
    }
```

```python
# проверка функций
data = [random.randint(1, 100) for _ in range(30)]
print("Список:", data)

freq = frequency_dict(data)
print("Частотный ряд:", freq)

stats = describe_list(data)
print("Статистики:", stats)
```
